```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score,classification_report
```

```python
data=pd.read_csv("IRIS.csv")
data
```

```python
data.dropna(inplace=True)
```

```python
x=data[['sepal_length','petal_length']]
y=data['species']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
# Kernal=linear
model=SVC(kernel="linear")
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

```python
acc=accuracy_score(y_test,y_pred)
acc
```

```python
cls=classification_report(y_test,y_pred)
print(cls)
```

```python
# kernal=rbf
models=SVC(kernel="rbf")
models.fit(x_train,y_train)
y_pre=models.predict(x_test)
```

```python
acc_score=accuracy_score(y_test,y_pre)
acc_score
```

```python
clss=classification_report(y_test,y_pre)
print(clss)
```

```python
# Kernal=poly
modelss=SVC(kernel="poly")
modelss.fit(x_train,y_train)
y_predi=modelss.predict(x_test)
```

```python
acc_scor=accuracy_score(y_test,y_predi)
acc_scor
```

```python
clas=classification_report(y_test,y_predi)
print(clas)
```
