```python
# Divisive Hierarchial Clustering
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import dendrogram,linkage
```

```python
data=pd.read_csv("/content/Mall_Customers.csv")
data
```

```python
data.info()
```

```python
data.describe()
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data.drop_duplicates()
```

```python
X=data[["Age","Annual Income (k$)"]]
X
```

```python
kmeans=KMeans(n_clusters=3,random_state=42,n_init=10)
```

```python
clust=kmeans.fit_predict(X)
data["Clusters"]=clust
```

```python
data
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.scatter(X['Age'],X['Annual Income (k$)'],c=clust)
plt.scatter(kmeans.cluster_centers_[:,0],
            kmeans.cluster_centers_[:,1],
            s=200,marker="o",color="Black")
plt.title("Divisive Clustering Scatter")
plt.xlabel("Age")
plt.ylabel("Annual Income")
plt.grid()
plt.show()
```

```python
L=linkage(X,method="complete")
plt.figure(figsize=(8,5))
dendrogram(L)
plt.title("Divisive Clustering Dendrogram")
plt.xlabel("Customers")
plt.ylabel("Euclidean Distance")
plt.grid(True)
plt.show()
```

```python
# Agglomerative Hierarchial Clustering
import numpy as np
import pandas as pd
from scipy.cluster.hierarchy import dendrogram,linkage
from sklearn.cluster import AgglomerativeClustering
```

```python
df=pd.read_csv("/content/Advertising Budget and Sales.csv")
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.head()
```

```python
df.columns
```

```python
df.rename(columns={'TV Ad Budget ($)':'TV Ad Budget'},inplace=True)
```

```python
df.rename(columns={"Radio Ad Budget ($)":"Radio Ad Budget",
                   'Newspaper Ad Budget ($)':'Newspaper Ad Budget',
                   'Sales ($)':'Sales'},inplace=True)
```

```python
df.columns
```

```python
df.drop_duplicates()
```

```python
df=df.drop('Unnamed: 0',axis=1)
df
```

```python
df.isnull().sum()
```

```python
x=df[['TV Ad Budget','Newspaper Ad Budget']]
x
```

```python
agg=AgglomerativeClustering(n_clusters=3)
clus=agg.fit_predict(x)
df['Clusters']=clus
```

```python
df
```

```python
plt.figure(figsize=(8,5))
plt.scatter(x['TV Ad Budget'],x['Newspaper Ad Budget'],c=clus)

plt.title("Agglomerative Clustering Scatter")
plt.xlabel("TV Ad Budget")
plt.ylabel("Newspaper Ad Budget")
plt.grid()
plt.show()
```

```python
a=linkage(x,method="ward")
plt.figure(figsize=(8,5))
dendrogram(a)
plt.title("Agglomerative Clustering Dendrogram")
plt.xlabel("Advertising Budget and Sales")
plt.ylabel("Euclidean Distance")
plt.grid(True)
plt.show()
```
