```python
import numpy as np
import pandas as pd
from statsmodels.tsa.arima.model import ARIMA
```

```python
data=pd.read_csv("continuous dataset.csv")
data
```

```python
data.info()
```

```python
data.columns
```

```python
data['date']=pd.to_datetime(data['datetime'])
data['date']
```

```python
data=data.sort_values('datetime')
data
```

```python
datas=data['T2M_toc']
```

```python
model=ARIMA(datas,order=(3,1,1))
model_fit=model.fit()
forecast=model_fit.forecast(steps=3)
```

```python
print("Day 1:", forecast.iloc[0])
print("Day 2:", forecast.iloc[1])
print("Day 3:", forecast.iloc[2])
```
