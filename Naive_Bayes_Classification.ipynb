```python
import pandas as pd
from sklearn.naive_bayes import MultinomialNB
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv("real_drug_dataset.csv")
df
```

```python
df.isnull().sum()
```

```python
le=LabelEncoder()
```

```python
df["Condition"]=le.fit_transform(df["Condition"])
df["Condition"]
```

```python
df["Drug_Name"]=le.fit_transform(df["Drug_Name"])
df["Drug_Name"]
```

```python
x=df [["Age","Condition","Drug_Name","Dosage_mg"]]
```

```python
y=le.fit_transform(df["Side_Effects"])
y
```

```python
models=MultinomialNB()
models.fit(x,y)
```

```python
test=pd.DataFrame({"Age":[45],"Condition":[1],"Drug_Name":[5],"Dosage_mg":[100]})
```

```python
pred=models.predict(test)
pred
```

```python
print("Predicted side effects:",le.inverse_transform(pred)[0])
```
