```python
text="Machine learning (ML) is a way to teach computers to learn from data and make decisions on their own, instead of following strict, pre-written rules."
answer=WordCloud(background_color="White")
answer.generate(text)
```

```python
import matplotlib.pyplot as plt
plt.imshow(answer)
plt.axis("off")
plt.show()
```
