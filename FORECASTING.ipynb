```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
```

```python
data=pd.read_csv("continuous dataset.csv")
data
```

```python
data.info()
```

```python
data.dtypes
```

```python
data['date']=pd.to_datetime(data['datetime'])
data['date']
```

```python
data=data.sort_values("datetime")
data
```

```python
datas=data['nat_demand']
```

```python
model=ARIMA(datas,order=(5,1,1))
model_fit= model.fit()
forecast=model_fit.forecast(steps=5)
```

```python
print("Next 5 Days Forecast:")
print("Day 1:", forecast.iloc[0])
print("Day 2:", forecast.iloc[1])
print("Day 3:", forecast.iloc[2])
print("Day 4:", forecast.iloc[3])
print("Day 5:", forecast.iloc[4])
```
